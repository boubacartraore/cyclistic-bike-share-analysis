```python
import pandas as pd

df = pd.read_csv(r"C:/Users/traor/Documents/Case_Study/Cyclistic_Capstone/data/cleaned/cyclistic_combined.csv")

print(df.shape)
```

```python
df["started_at"] = pd.to_datetime(df["started_at"])
df["ended_at"] = pd.to_datetime(df["ended_at"])
```

```python
df.info()
```

```python
df["ride_length"] = (
    df["ended_at"] -
    df["started_at"]
).dt.total_seconds() / 60
```

```python
(df["ride_length"] <= 0).sum()
```

```python
df = df[df["ride_length"] > 0]
```

```python
df = df.drop_duplicates(
    subset="ride_id"
)
```

```python
df["day_of_week"] = df["started_at"].dt.day_name()
```

```python
df["month"] = df["started_at"].dt.month_name()
```

```python
df["hour"] = df["started_at"].dt.hour
```

```python
df["year"] = df["started_at"].dt.year
```

```python
df["is_weekend"] = df["day_of_week"].isin(
    ["Saturday", "Sunday"]
)
```

```python
df.to_csv(
    r"C:/Users/traor/Documents/Case_Study/Cyclistic_Capstone/data/cleaned/cyclistic_cleaned.csv",
    index=False
)
```
