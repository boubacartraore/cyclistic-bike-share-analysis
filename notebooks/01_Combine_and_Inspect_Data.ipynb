```
import pandas as pd
import zipfile
from pathlib import Path
data_folder = Path(r"C:/Users/traor/Documents/Case_Study/Cyclistic_Capstone/data/raw")
zip_files = sorted(data_folder.glob("*.zip"))

print(f"Number of ZIP files found: {len(zip_files)}")

for file in zip_files:
    print(file.name)
```

```
dataframes = []

for zip_file in zip_files:

    with zipfile.ZipFile(zip_file, "r") as z:

        # Ignore macOS hidden files
        csv_name = [f for f in z.namelist()
                    if f.endswith(".csv") and "__MACOSX" not in f][0]

        df = pd.read_csv(z.open(csv_name))

        print(f"{csv_name}: {df.shape}")

        dataframes.append(df)
```

```
columns_match = all(
    df.columns.equals(dataframes[0].columns)
    for df in dataframes
)

print("All files have identical columns:", columns_match)
```

```
combined_df = pd.concat(
    dataframes,
    ignore_index=True
)
```

```
print("Shape:", combined_df.shape)

combined_df.head()
```

```
cleaned_folder = Path(r"C:/Users/traor/Documents/Case_Study/Cyclistic_Capstone/data/cleaned")
cleaned_folder.mkdir(parents=True, exist_ok=True)
```

```
combined_df.to_csv(
    cleaned_folder / "cyclistic_combined.csv",
    index=False
)

print("Combined dataset saved successfully!")
```
